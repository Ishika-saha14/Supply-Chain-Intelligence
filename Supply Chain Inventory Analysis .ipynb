```python
print("SKU IDs:", df['SKU_ID'].nunique())
print("Warehouses:", df['Warehouse_ID'].nunique())
print("Suppliers:", df['Supplier_ID'].nunique())
print("Regions:", df['Region'].nunique())

print(df['Region'].unique())
```

```python
#Total Units Sold
total_units_sold = df['Units_Sold'].sum()
print("Total Units Sold:", total_units_sold)
```

```python
#Average Daily Units Sold
daily_sales = df.groupby('Date')['Units_Sold'].sum()
avg_daily_sales = daily_sales.mean()
print("Average Daily Units Sold:", avg_daily_sales)
```

```python
# Inventory Units on Latest Recorded Date

latest_date = df['Date'].max()
latest_inventory = df[df['Date'] == latest_date]

total_inventory = latest_inventory['Inventory_Level'].sum()

print("Latest Recorded Date:", latest_date)
print("Inventory Units:", total_inventory)
```

```python
# Inventory Value on Latest Recorded Date

df['Inventory_Value'] = (
    df['Inventory_Level'] * df['Unit_Cost']
)

total_inventory_value = (
    latest_inventory['Inventory_Level'] *
    latest_inventory['Unit_Cost']
).sum()

print(
    "Inventory Value on Latest Recorded Date:",
    round(total_inventory_value, 2)
)
```

```python
#Sales by Region
sales_by_region = df.groupby('Region')['Units_Sold'].sum()
print(sales_by_region)
```

```python
import matplotlib.pyplot as plt

sales_by_region.plot(kind='bar')

plt.title('Units Sold by Region')
plt.xlabel('Region')
plt.ylabel('Units Sold')
plt.ylim(450000,465000)


plt.show()
```

```python
#Sales by SKU
sales_by_sku = df.groupby('SKU_ID')['Units_Sold'].sum()
print(sales_by_sku.sort_values(ascending = False).head(10))
```

```python
import matplotlib.pyplot as plt
top_10_skus = sales_by_sku.sort_values(ascending=False).head(10)

top_10_skus.plot(kind='bar', title='Top 10 SKUs by Units Sold')

plt.ylabel('Units Sold')
plt.xlabel('SKU ID')
plt.ylim(36000, 37500)

plt.show()
```

```python
#Inventory by Warehouse
inventory_by_warehouse = df.groupby('Warehouse_ID')['Inventory_Level'].sum()
print(inventory_by_warehouse)
```

```python
import matplotlib.pyplot as plt

inventory_by_warehouse.plot(kind='bar')

plt.title('Inventory by Warehouse')
plt.xlabel('Warehouse')
plt.ylabel('Inventory Units')
plt.ylim(8_200_000, 9_100_000)

plt.show()
```

```python
#Inventory vs Reorder Point
low_inventory = df[df['Inventory_Level'] <= df['Reorder_Point']]
print("Records at or below reorder point:", len(low_inventory))
```

```python
#low Inventory by Warehouse
low_inventory_by_warehouse = low_inventory.groupby('Warehouse_ID').size()
print(low_inventory_by_warehouse)
```

```python
import matplotlib.pyplot as plt

low_inventory_by_warehouse.plot(kind='bar')

plt.title('Low Inventory Records by Warehouse')
plt.xlabel('Warehouse')
plt.ylabel('Number of Records')
plt.ylim(990, 1025)

plt.show()
```

```python
# Sales Trend Over Time

daily_sales = df.groupby('Date')['Units_Sold'].sum()
daily_sales.plot()
plt.title('Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Units Sold')

plt.show()
```

```python
# Inventory Trend Over Time

daily_inventory = df.groupby('Date')['Inventory_Level'].mean()
daily_inventory.plot()
plt.title('Inventory Trend Over Time')
plt.xlabel('Date')
plt.ylabel('Average Inventory')

plt.show()
```

```python
# Revenue Analysis

df['Revenue'] = df['Units_Sold'] * df['Unit_Price']
total_revenue = df['Revenue'].sum()
print("Total Revenue:", total_revenue)
```

```python
# Promotion vs Sales

promotion_sales = df.groupby('Promotion_Flag')['Units_Sold'].mean()
print(promotion_sales)
```

```python
from getpass import getpass
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

password = getpass("Enter MySQL password: ")

connection_url = URL.create(
    "mysql+pymysql",
    username="root",
    password=password,
    host="localhost",
    port=3306,
    database="supply_chain"
)

engine = create_engine(connection_url)

print("Engine created successfully!")
```
